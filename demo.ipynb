## 1. Setup

First, we import our custom `mininumpy` library, the official `numpy` library for comparison, and `matplotlib` for data visualization.

```python
import numpy as np
import mininumpy as mnp
import math
import matplotlib.pyplot as plt

# Set print options for NumPy for better readability
np.set_printoptions(precision=4, suppress=True)
```

## 2. Core Functionality: Array Creation & Manipulation

We will test the array creation functions of `mininumpy` and compare their output with `numpy`.

```python
print("--- 1. Create array from list ---")
list_data = [[1, 2, 3], [4, 5, 6]]
mnp_a = mnp.array(list_data)
np_a = np.array(list_data)
print("MiniNumPy:\n", mnp_a)
print("NumPy:\n", np_a)

print("\n--- 2. Create array of zeros ---")
mnp_zeros = mnp.zeros((2, 3))
np_zeros = np.zeros((2, 3))
print("MiniNumPy:\n", mnp_zeros)
print("NumPy:\n", np_zeros)

print("\n--- 3. Create array of ones ---")
mnp_ones = mnp.ones((3, 2))
np_ones = np.ones((3, 2))
print("MiniNumPy:\n", mnp_ones)
print("NumPy:\n", np_ones)

print("\n--- 4. Create identity matrix ---")
mnp_eye = mnp.eye(3)
np_eye = np.eye(3)
print("MiniNumPy:\n", mnp_eye)
print("NumPy:\n", np_eye)

print("\n--- 5. Create arange array ---")
mnp_arange = mnp.arange(0, 10, 2)
np_arange = np.arange(0, 10, 2)
print("MiniNumPy:\n", mnp_arange)
print("NumPy:\n", np_arange)

print("\n--- 6. Create linspace array ---")
mnp_linspace = mnp.linspace(0, 1, 5)
np_linspace = np.linspace(0, 1, 5)
print("MiniNumPy:\n", mnp_linspace)
print("NumPy:\n", np_linspace)
```

### 2.1 Array Attributes and Basic Manipulation

```python
print("--- Array Attributes ---")
print(f"Shape: {mnp_a.shape}, Expected: {np_a.shape}")
print(f"ndim: {mnp_a.ndim}, Expected: {np_a.ndim}")
print(f"Size: {mnp_a.size}, Expected: {np_a.size}")

print("\n--- Reshape Operation ---")
mnp_reshaped = mnp_a.reshape((3, 2))
np_reshaped = np_a.reshape((3, 2))
print("MiniNumPy Reshaped:\n", mnp_reshaped)
print("NumPy Reshaped:\n", np_reshaped)

print("\n--- Transpose Operation ---")
mnp_transposed = mnp_a.transpose()
np_transposed = np_a.transpose()
print("MiniNumPy Transposed:\n", mnp_transposed)
print("NumPy Transposed:\n", np_transposed)
```

## 3. Elementwise Operations

Testing element-by-element arithmetic, mathematical functions, and reduction functions.

```python
mnp_b = mnp.array([[10, 20, 30], [40, 50, 60]])
np_b = np.array([[10, 20, 30], [40, 50, 60]])

print("--- Arithmetic Operations ---")
print("Addition (mnp):\n", mnp_a + mnp_b)
print("Addition (np):\n", np_a + np_b)

print("\nScalar Multiplication (mnp):\n", mnp_a * 10)
print("Scalar Multiplication (np):\n", np_a * 10)

print("\n--- Mathematical Functions ---")
mnp_c = mnp.array([1, 4, 9])
np_c = np.array([1, 4, 9])
print("Sqrt (mnp):\n", mnp.sqrt(mnp_c))
print("Sqrt (np):\n", np.sqrt(np_c))

print("\n--- Reduction Functions ---")
print(f"Sum (mnp): {mnp.sum(mnp_a)}, Expected: {np.sum(np_a)}")
print(f"Mean (mnp): {mnp.mean(mnp_a)}, Expected: {np.mean(np_a)}")
print(f"Max (mnp): {mnp.max(mnp_a)}, Expected: {np.max(np_a)}")
```

## 4. Linear Algebra Module (`minilinalg`)

Testing core linear algebra functionalities like matrix multiplication, determinant, and inverse.

```python
mnp_m1 = mnp.array([[1, 2], [3, 4]])
mnp_m2 = mnp.array([[5, 6], [7, 8]])
np_m1 = np.array([[1, 2], [3, 4]])
np_m2 = np.array([[5, 6], [7, 8]])

print("--- Matrix Multiplication (@) ---")
mnp_matmul = mnp_m1 @ mnp_m2
np_matmul = np_m1 @ np_m2
print("MiniNumPy Matmul:\n", mnp_matmul)
print("NumPy Matmul:\n", np_matmul)

print("\n--- Determinant (det) ---")
# Using a slightly more complex 3x3 matrix
mnp_m3 = mnp.array([[1, 2, 3], [0, 4, 5], [1, 0, 6]])
np_m3 = np.array([[1, 2, 3], [0, 4, 5], [1, 0, 6]])
mnp_det = mnp.linalg.det(mnp_m3)
np_det = np.linalg.det(np_m3)
print(f"Determinant (mnp): {mnp_det:.4f}")
print(f"Determinant (np):  {np_det:.4f}")

print("\n--- Matrix Inverse (inv) ---")
mnp_inv = mnp.linalg.inv(mnp_m3)
np_inv = np.linalg.inv(np_m3)
print("MiniNumPy Inverse:\n", mnp_inv)
print("NumPy Inverse:\n", np_inv)

print("\n--- Advanced Linear Algebra (5x5 Matrix) ---")
# Generate a 5x5 matrix using list comprehension
# Using a pattern ensuring it's likely invertible (diagonally dominant-ish)
data_5x5 = [
    [5, 1, 0, 2, 1],
    [1, 6, 2, 0, 1],
    [0, 2, 5, 1, 0],
    [2, 0, 1, 6, 2],
    [1, 1, 0, 2, 5]
]

mnp_big = mnp.array(data_5x5)
np_big = np.array(data_5x5)

print(f"Matrix Shape: {mnp_big.shape}")

# 1. Determinant
# Note: For 5x5, recursive method would be slow (5! operations),
# but our LU method is fast (n^3).
det_mnp = mnp.linalg.det(mnp_big)
det_np = np.linalg.det(np_big)

print(f"\nDeterminant (mnp): {det_mnp:.4f}")
print(f"Determinant (np):  {det_np:.4f}")

# 2. Inverse
inv_mnp = mnp.linalg.inv(mnp_big)
print("\nInverse Matrix (mnp):\n", inv_mnp)

# 3. Verification: A @ A_inv should be Identity
identity_check = mnp_big @ inv_mnp
print("\nVerification (A @ A_inv ≈ Identity):\n", identity_check)

# 4. Check error against NumPy
inv_np = np.linalg.inv(np_big)
# Calculate simple difference sum
diff = 0
for i in range(5):
    for j in range(5):
        diff += abs(inv_mnp.data[i][j] - inv_np[i][j])
print(f"\nTotal Error vs NumPy: {diff:.8f}")
```

## 5. Practical Applications

Demonstrating `mininumpy` in two mini-projects: 2D geometric transformations and simple image manipulation.

### 5.1 Application 1: 2D Geometric Transformations

We will define a simple shape (the outline of a house), apply scaling, rotation, and shear transformations using matrix multiplication, and visualize the results with `matplotlib`.

```python
# Define a simple house shape with (x, y) coordinates
# We use homogeneous coordinates (x, y, 1) for matrix transformations
house_coords = mnp.array([
    [0, 0, 1],      # 1. Bottom-left (Start)
    [1, 0, 1],      # 2. Bottom-right
    [1, 1, 1],      # 3. Top-right
    [0.5, 1.5, 1],  # 4. Roof tip
    [0, 1, 1],      # 5. Top-left
    [0, 0, 1]       # 6. Back to Bottom-left (Close shape)
])

# Define a plotting helper function
def plot_shape(coords, title):
    x_values = [row[0] for row in coords.data]
    y_values = [row[1] for row in coords.data]

    plt.plot(x_values, y_values)

    plt.title(title)
    plt.xlim(-2, 2)
    plt.ylim(-2, 2)
    plt.gca().set_aspect('equal', adjustable='box')
    plt.grid(True)

# --- 1. Define Transformation Matrices ---

# Scaling matrix: 1.5x in x-axis, 0.5x in y-axis
scale_matrix = mnp.array([
    [1.5, 0, 0],
    [0, 0.5, 0],
    [0, 0, 1]
])

# Rotation matrix: Rotate counter-clockwise by 45 degrees
angle = math.radians(45)
c, s = math.cos(angle), math.sin(angle)
rotation_matrix = mnp.array([
    [c, s, 0],
    [-s, c, 0],
    [0, 0, 1]
])

# Shear matrix: Shear along the x-axis
shear_matrix = mnp.array([
    [1, 0.5, 0],
    [0, 1, 0],
    [0, 0, 1]
])

# --- 2. Apply Transformations ---
scaled_house = house_coords @ scale_matrix
rotated_house = house_coords @ rotation_matrix
sheared_house = house_coords @ shear_matrix

# --- 3. Visualize the Results ---
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plot_shape(house_coords, 'Original House')

plt.subplot(2, 2, 2)
plot_shape(scaled_house, 'Scaled House')

plt.subplot(2, 2, 3)
plot_shape(rotated_house, 'Rotated House (45 deg)')

plt.subplot(2, 2, 4)
plot_shape(sheared_house, 'Sheared House')

plt.show()
```

### 5.2 Application 2: Simple Image Processing

We will create a simple 8x8 grayscale image (represented by a `mininumpy` matrix) and perform two basic operations on it: brightness enhancement and color inversion.

```python
# Create an 8x8 grayscale image (0=black, 255=white)
image = mnp.zeros((9, 9))
# Draw a bright cross on the image
image.data[4][:] = [180] * 9
for i in range(9):
    image.data[i][4] = 180

# --- 1. Increase Brightness ---
# Achieved by adding a constant value to every pixel
brighter_image = image + 50
# Note: In a real application, pixel values > 255 would need to be clipped.
# This is simplified for demonstration purposes.

# --- 2. Invert Colors ---
# Achieved by subtracting the current pixel value from the maximum value
max_pixel_value = 255
inverted_image = max_pixel_value - image

# --- 3. Visualize the Results ---
plt.figure(figsize=(15, 5))
ticks = range(9)

plt.subplot(1, 3, 1)
plt.imshow(image.data, cmap='gray', vmin=0, vmax=255)
plt.title('Original Image')
plt.xticks(ticks)
plt.yticks(ticks)

plt.subplot(1, 3, 2)
plt.imshow(brighter_image.data, cmap='gray', vmin=0, vmax=255)
plt.title('Brighter Image')
plt.xticks(ticks)
plt.yticks(ticks)

plt.subplot(1, 3, 3)
plt.imshow(inverted_image.data, cmap='gray', vmin=0, vmax=255)
plt.title('Inverted Image')
plt.xticks(ticks)
plt.yticks(ticks)

plt.show()
```

## 6. Conclusion

In this project, we successfully designed and implemented `MiniNumPy`, a lightweight numerical computing library built from scratch in pure Python. The library replicates the core architecture of NumPy and achieves the following key milestones:

### 1. Robust Core Architecture
* Implemented an N-dimensional **`Array` class** supporting flattening, reshaping, and broadcasting mechanisms.
* Overloaded Python operators (`+`, `-`, `*`, `@`) to enable intuitive, mathematical syntax for array manipulations.

### 2. Advanced Linear Algebra with LU Decomposition
* Instead of using the computationally expensive recursive approach ($O(n!)$) for determinants and inverses, we implemented **LUP Decomposition with Partial Pivoting** ($PA = LU$).
* This optimization reduces the time complexity to **$O(n^3)$**, allowing our library to handle larger matrices (e.g., $5 \times 5$ or larger) efficiently and numerically stably.
* We verified the precision of our `inv()` and `det()` functions against NumPy, achieving negligible error margins.

### 3. Practical Applications
* **2D Transformations**: Successfully applied matrix multiplication to perform scaling, rotation, and shearing on geometric shapes, correctly handling homogeneous coordinates.
* **Image Processing**: Demonstrated matrix-based image manipulation (brightness adjustment and color inversion), verifying the correctness of the underlying array arithmetic.

### Summary
While `MiniNumPy` cannot match the execution speed of NumPy (which relies on optimized C/Fortran backends like BLAS/LAPACK), it serves as an excellent educational tool. By building it, we gained a deep understanding of memory layout, broadcasting rules, and the algorithmic foundations of modern numerical computing.
